```python
# ── Cell 2: storage setup on the USB drive (D:) ──
import sys, shutil, json, importlib
from pathlib import Path

DATA_ROOT = Path(r"D:\kalshi_demand")

# 1. Is the drive there, and how much room is on it?
drive = Path(DATA_ROOT.anchor)
assert drive.exists(), "Drive D: not found. Check the USB drive is plugged in."
total, used, free = shutil.disk_usage(drive)
print(f"Python {sys.version.split()[0]} | Drive D: {free/1e9:.1f} GB free of {total/1e9:.1f} GB")

# 2. Folder layout
#    raw/       = API responses exactly as received; never edited
#    interim/   = cleaned and joined tables
#    processed/ = model-ready datasets
#    logs/      = crawl progress, so interrupted pulls can resume
for sub in ["raw/series", "raw/markets", "raw/candles", "interim", "processed", "logs"]:
    (DATA_ROOT / sub).mkdir(parents=True, exist_ok=True)

# 3. Prove we can write and read back
test = DATA_ROOT / "logs" / "_write_test.json"
test.write_text(json.dumps({"ok": True}))
print("Write/read test:", json.loads(test.read_text()))
test.unlink()

# 4. Package check
for pkg in ["pandas", "pyarrow", "requests"]:
    m = importlib.import_module(pkg)
    print(f"  {pkg} {m.__version__}")

print("\nFolders on drive:")
for p in sorted(DATA_ROOT.rglob("*")):
    if p.is_dir():
        print("  ", p.relative_to(DATA_ROOT))
```
